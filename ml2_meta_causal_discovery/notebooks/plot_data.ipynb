```python
from ml2_meta_causal_discovery.datasets.dataset_generators import \
    ClassifyDatasetGenerator


dataset_generator = ClassifyDatasetGenerator(
        num_variables=2,
        function_generator="gp",
        batch_size=100,
        num_samples=2000,
        kernel_sum=True,
        mean_function="latent",
        graph_type=["ER"],
        graph_degrees=[0, 1, 2],
    )
```

```python
(
    target_data,
    causal_graphs,
) = next(dataset_generator.generate_next_dataset())
```

```python
import matplotlib.pyplot as plt

target_data = (target_data - target_data.mean(axis=1, keepdims=True)) / target_data.std(axis=1, keepdims=True)
for i in range(100):
    plt.figure(figsize=(15, 5))
    print(causal_graphs[i])
    for j in range(2):
        x = 0 if j == 0 else 1
        y = 1 if j == 0 else 0
        plt.scatter(target_data[i, :, x], target_data[i, :, y])
        plt.title(f'Variable {j + 1}')
    plt.show()
```
